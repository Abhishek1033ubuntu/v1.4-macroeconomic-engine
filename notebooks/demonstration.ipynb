# V1.4 Macroeconomic Engine: Demonstration & Sensitivity Analysis
This notebook demonstrates how to load the V1.4 Macroeconomic Engine, execute a 5-year simulation run, and visualize the net treasury position.

```python
import sys
sys.path.append('..')

import matplotlib.pyplot as plt
from src.engine import V14MasterEconomyModel
from src.sectors import (
    fmcg_formal, fmcg_informal, liquor_excise, 
    energy_fuel, real_estate, fisheries
)

# 1. Initialize Engine
model = V14MasterEconomyModel(years=5, gdp_growth_rate=0.065)

# 2. Register Sectors
for sector in [fmcg_formal, fmcg_informal, liquor_excise, energy_fuel, real_estate, fisheries]:
    model.register_sector(sector)

# 3. Run Simulation
df = model.run_simulation()
df.tail()
```

```python
# Plot Cumulative Net Position
plt.figure(figsize=(10, 4))
plt.plot(df['Month'], df['V14_Net_Position'] / 1000, label='V1.4 Net Surplus Trajectory', color='green', linewidth=2)
plt.plot(df['Month'], df['Trad_Net_Position'] / 1000, label='Traditional GST Deficit Trajectory', color='red', linestyle='--')
plt.axhline(0, color='gray', linestyle=':')
plt.title('5-Year Net Treasury Position (Crore INR)')
plt.xlabel('Month')
plt.ylabel('Crore INR')
plt.legend()
plt.grid(True)
plt.show()
```
